```python
data_melb_airport = data.loc[data["Location"]=="MelbourneAirport",["Date","MaxTemp"]]
```

```python
data_melb_airport = (
    data_melb_airport
    .sort_values("Date")
    .reset_index(drop=True)
)
```

```python
print(
    "Duplicate dates:",
    data_melb_airport["Date"].duplicated().sum()
)

expected_dates = pd.date_range(
    start=data_melb_airport["Date"].min(),
    end=data_melb_airport["Date"].max(),
    freq="D"
)

missing_dates = expected_dates.difference(
    data_melb_airport["Date"]
)

print("Missing calendar dates:", len(missing_dates))
print(missing_dates)
```
